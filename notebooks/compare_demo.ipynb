# Music similarity demo

Compares the melodic/harmonic similarity between two music files (any mix of MP3, WAV, MIDI, ...).

```python
import librosa.display
import matplotlib.pyplot as plt

import musicsim as ms

PATH_A = "../data/waka.mp3"
PATH_B = "../data/bop.mp3"
```

```python
result = ms.compare(PATH_A, PATH_B)
print(f"similarity score: {result.score:.3f}  (avg alignment cost: {result.cost:.3f})")

#0.948 body - bop
#0.972 waka - zamina
```

```python
fig, axes = plt.subplots(2, 1, figsize=(10, 6))
librosa.display.specshow(result.chroma_a, y_axis="chroma", x_axis="time", ax=axes[0])
axes[0].set_title("A: chroma")
librosa.display.specshow(result.chroma_b, y_axis="chroma", x_axis="time", ax=axes[1])
axes[1].set_title("B: chroma")
fig.tight_layout()
```

```python
path = result.alignment_path
plt.figure(figsize=(6, 6))
plt.plot(path[:, 0], path[:, 1])
plt.xlabel("A frame index")
plt.ylabel("B frame index")
plt.title("DTW alignment path")
plt.gca().set_aspect("equal")
```
